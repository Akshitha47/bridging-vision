```python
# move 0.1% of files from one directory to another
import os
import random
import shutil

def move_files(src, dst, percentage):
    if not os.path.exists(dst):
        os.makedirs(dst)
    files = os.listdir(src)
    num_files = len(files)
    num_to_move = int(num_files * percentage)
    to_move = random.sample(files, num_to_move)
    for f in to_move:
        shutil.move(os.path.join(src, f), os.path.join(dst, f))


SOURCE = "Flicker8k_Dataset"
DESTINATION = "Flicker8k_Dataset_test"
PERCENTAGE = 0.001
move_files(SOURCE, DESTINATION, PERCENTAGE)
```

```python
def load_references(file_path):
    references = {}
    with open(file_path, 'r') as file:
        for line in file:
            image_id = line.split()[0][:-2]
            caption = line.split()[1:]
            caption = ' '.join(caption)
            if image_id not in references:
                references[image_id] = []
            references[image_id].append(caption)
    return references

references = load_references('Flickr8k.token.test.txt')
references
```
